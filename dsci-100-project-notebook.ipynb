#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
#We need to tidy the data for the research question
march_madness_tidy <- march_madness |>
#Since the research question only takes into account seed, regional-season win percentage, and tournament finish, those are the only columns we are keeping.
#seed and regional-season win percentage are the two numeric predictors, and tournament finish is the categorial response.
    select(seed, reg_wins_pct, tourney_finish) |>
#We need to remove rows with missing values in any of these columns (incase they have NA)
    filter(!is.na(seed), !is.na(reg_wins_pct), !is.na(tourney_finish)) |>
#As mentioned in the description of the variables, the 1983 game had teams who were "seed 0"
#These teams were competing in the opening round game to become the #8 seed. The losers of this opening round game was eliminated and had no seed.
#In this data set, a lower seed means a stronger team. A 1 seed is a top team, and a 16 seed is a weak one. 
#k-nn treats seed as a number and finds neighbours by distance, so it would read a seed of 0 as even stronger than a 1 seed.
#This is an error as these teams were actually among the weaker ones in the 1983 tournament, playing an extra game just to earn an 8 seed. 
#Leaving them as 0 would give the model misleading data and could distort its predictions.
#Thus, we need to remove them. There are only a handul of seed 0 teams.
#Removing them would not majorly impact the data
    filter(seed != 0) |>
#We need to convert the tourney_finish into a factor as it is categorical. It was previously chr.
    mutate(tourney_finish = as_factor(tourney_finish))
#There's no need to pivot, because no variable is spread across several columns and no column mixes two variables.
#Since we only want 6 rows of the tidy data, we can use head() function.
head(march_madness_tidy)
           



```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
#This line tells jupyter to display all the columns.
options(repr.matrix.max.cols = 50)
# Each line inside summarize() makes one new column. new_name = calculation.
# mean() gives the average, min() the smallest and max() the largest value.
# na.rm = TRUE drops missing values first.
march_madness_summary <- march_madness |>
  summarize(
    # Year of the tournament
    year_mean = mean(year, na.rm = TRUE),
    year_min = min(year, na.rm = TRUE),
    year_max = max(year, na.rm = TRUE),
    # Tournament seed (includes the 1983 seed 0 teams here)
    seed_mean = mean(seed, na.rm = TRUE),
    seed_min = min(seed, na.rm = TRUE),
    seed_max = max(seed, na.rm = TRUE),
    # Conference wins
    conf_wins_mean = mean(conf_wins, na.rm = TRUE),
    conf_wins_min = min(conf_wins, na.rm = TRUE),
    conf_wins_max = max(conf_wins, na.rm = TRUE),
    # Conference losses
    conf_losses_mean = mean(conf_losses, na.rm = TRUE),
    conf_losses_min = min(conf_losses, na.rm = TRUE),
    conf_losses_max = max(conf_losses, na.rm = TRUE),
    # Conference win percentage
    conf_wins_pct_mean = mean(conf_wins_pct, na.rm = TRUE),
    conf_wins_pct_min = min(conf_wins_pct, na.rm = TRUE),
    conf_wins_pct_max = max(conf_wins_pct, na.rm = TRUE),
    # Rank within the conference
    conf_rank_mean = mean(conf_rank, na.rm = TRUE),
    conf_rank_min = min(conf_rank, na.rm = TRUE),
    conf_rank_max = max(conf_rank, na.rm = TRUE),
    # Regular-season wins
    reg_wins_mean = mean(reg_wins, na.rm = TRUE),
    reg_wins_min = min(reg_wins, na.rm = TRUE),
    reg_wins_max = max(reg_wins, na.rm = TRUE),
    # Regular-season losses
    reg_losses_mean = mean(reg_losses, na.rm = TRUE),
    reg_losses_min = min(reg_losses, na.rm = TRUE),
    reg_losses_max = max(reg_losses, na.rm = TRUE),
    # Regular-season win percentage
    reg_wins_pct_mean = mean(reg_wins_pct, na.rm = TRUE),
    reg_wins_pct_min = min(reg_wins_pct, na.rm = TRUE),
    reg_wins_pct_max = max(reg_wins_pct, na.rm = TRUE),
    # Tournament wins
    tourney_wins_mean = mean(tourney_wins, na.rm = TRUE),
    tourney_wins_min = min(tourney_wins, na.rm = TRUE),
    tourney_wins_max = max(tourney_wins, na.rm = TRUE),
    # Tournament losses
    tourney_losses_mean = mean(tourney_losses, na.rm = TRUE),
    tourney_losses_min = min(tourney_losses, na.rm = TRUE),
    tourney_losses_max = max(tourney_losses, na.rm = TRUE),
    # Total wins (season plus tournament)
    total_wins_mean = mean(total_wins, na.rm = TRUE),
    total_wins_min = min(total_wins, na.rm = TRUE),
    total_wins_max = max(total_wins, na.rm = TRUE),
    # Total losses
    total_losses_mean = mean(total_losses, na.rm = TRUE),
    total_losses_min = min(total_losses, na.rm = TRUE),
    total_losses_max = max(total_losses, na.rm = TRUE),
    # Total win percentage
    total_wins_pct_mean = mean(total_wins_pct, na.rm = TRUE),
    total_wins_pct_min = min(total_wins_pct, na.rm = TRUE),
    total_wins_pct_max = max(total_wins_pct, na.rm = TRUE))

#Show the table
march_madness_summary
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
# is.na(x) gives TRUE for each missing value, and sum() counts the TRUEs, because R treats TRUE as 1 and FALSE as 0.
# is.na() works on numeric (<dbl>) and categorical (<chr>) columns alike.
march_madness_missing <- march_madness |>
  # summarize() collapses the data into one row of counts and one column per variable.
  summarize(
    year = sum(is.na(year)),
    school = sum(is.na(school)),
    seed = sum(is.na(seed)),
    conference = sum(is.na(conference)),
    conf_wins = sum(is.na(conf_wins)),
    conf_losses = sum(is.na(conf_losses)),
    conf_wins_pct = sum(is.na(conf_wins_pct)),
    conf_rank = sum(is.na(conf_rank)),
    division = sum(is.na(division)),
    reg_wins = sum(is.na(reg_wins)),
    reg_losses = sum(is.na(reg_losses)),
    reg_wins_pct = sum(is.na(reg_wins_pct)),
    bid = sum(is.na(bid)),
    first_game_at_home = sum(is.na(first_game_at_home)),
    tourney_wins = sum(is.na(tourney_wins)),
    tourney_losses = sum(is.na(tourney_losses)),
    tourney_finish = sum(is.na(tourney_finish)),
    total_wins = sum(is.na(total_wins)),
    total_losses = sum(is.na(total_losses)),
    total_wins_pct = sum(is.na(total_wins_pct)))

#show output
march_madness_missing
```

#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

- Use `ggplot` to plot the distribution of one of the numeric variables in your tidied data set.

- Use `ggplot` to create a visualization showing the relative proportions of each category/level for one of categorical variable in your tidied data set. 

- Clearly label the variable that is being plotted in each plot.

*Notes:* 
- *Be sure the output is as clear as possible, provide an English title along with readable x and y column names*

- *You may also add more code cells to this section*

</div>

```r
# Plot 1 is the distribution of the numeric variable reg_wins_pct (part of the research question).
# A histogram shows how many teams fall into each range of win percentage.
reg_wins_pct_plot <- march_madness_tidy |>
  # aes(x = reg_wins_pct) puts win percentage as the x axis.
  ggplot(aes(x = reg_wins_pct)) +
  # geom_histogram() draws the bars; binwidth = 2.5 groups teams into 2.5% ranges
  # color = "white" draws thin lines between bars so they're easy to tell apart
  geom_histogram(binwidth = 2.5, color = "white") +
  # labs() adds title and axis names.
  labs(title = "Distribution of Regular-Season Win Percentage",
       x = "Regular-Season Win Percentage (%)",
       y = "Number of Teams") +
  # theme(text = ...) makes all the text larger so it's easy to read.
  theme(text = element_text(size = 14))

# Show the plot
reg_wins_pct_plot
```

```r
finish_proportions <- march_madness_tidy |>
  # count() gives the number of teams in each tournament finish (column n).
  count(tourney_finish) |>
  # mutate() adds a proportion column: each count divided by the total number of teams.
  mutate(proportion = n / sum(n))

finish_plot <- finish_proportions |>
  # Finish goes on the x-axis and its proportion on the y-axis.
  ggplot(aes(x = tourney_finish, y = proportion)) +
  # geom_col() draws one bar per finish, with height equal to its proportion.
  geom_col() +
  # labs() adds a plain-English title and readable axis names.
  labs(title = "Proportion of Teams by Tournament Finish",
       x = "Tournament Finish",
       y = "Proportion of Teams") +
  # Larger text, and x labels tilted 45 degrees so the long names don't overlap.
  theme(text = element_text(size = 14),
        axis.text.x = element_text(angle = 45, hjust = 1))

# Show the plot.
finish_plot
```
