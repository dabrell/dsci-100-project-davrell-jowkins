```r
# set.seed() makes the random split and folds the same every time we run the code.
set.seed(2024)



# initial_split() randomly splits the tidy data to 75% training and 25% testing.
# strata = tourney_finish keeps each finish in similar proportions in both sets. Ensuring no lopsided data. 
# This matters because some finishes (like champ) are rare.
mm_split <- initial_split(march_madness_tidy, prop = 0.75, strata = tourney_finish)

# training() and testing() pull out the two sets.
mm_train <- training(mm_split)
mm_test <- testing(mm_split)
```

```r
# The recipe says which variables to use and how to prepare them
# tourney_finish ~ seed + reg_wins_pct means predict finish from seed and win percentage.
# data = mm_train means the scaling is learned from training data only.
mm_recipe <- recipe(tourney_finish ~ seed + reg_wins_pct, data = mm_train) |>
  # step_scale() and step_center() put both predictors on the same scale,
  # so win percentage doesn't outweigh seed in euclidean distance. 
  step_scale(all_predictors()) |>
  step_center(all_predictors())

# KNN classification, with K left as tune() so we can pick the best one
# weight_func = "rectangular" means every neighbour gets an equal vote
knn_tune_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = tune()) |>
  set_engine("kknn") |>
  set_mode("classification")
```

```r
# vfold_cv() splits the training set into 5 folds.stratified by finish.
mm_vfold <- vfold_cv(mm_train, v = 5, strata = tourney_finish)

# The K values to try is 1 to 50 in steps of 2. This is done as it is a large range of K values.
k_vals <- tibble(neighbors = seq(from = 1, to = 50, by = 2))

# tune_grid() fits the model for every K on every fold and records accuracy.
knn_results <- workflow() |>
  add_recipe(mm_recipe) |>
  add_model(knn_tune_spec) |>
  tune_grid(resamples = mm_vfold, grid = k_vals) |>
  # collect_metrics() gathers the average results across the 5 folds.
  collect_metrics()

# We need to only keep accuracy, since that's the metric we're using.
accuracies <- knn_results |>
  filter(.metric == "accuracy")

# We need to plot accuracy against K to see which K works best
accuracy_vs_k <- accuracies |>
  ggplot(aes(x = neighbors, y = mean)) +
  geom_point() +
  geom_line() +
  labs(title = "Cross-Validation Accuracy for Each K",
       x = "Number of Neighbours (K)",
       y = "Mean Accuracy (5-fold CV)") +
  theme(text = element_text(size = 14))

accuracy_vs_k

# Then, we pick the K with the highest mean accuracy.
best_k <- accuracies |>
  slice_max(mean, n = 1, with_ties = FALSE) |>
  pull(neighbors)

best_k
```

```r
# Build the final model using the best K found above.
knn_spec <- nearest_neighbor(weight_func = "rectangular", neighbors = best_k) |>
  set_engine("kknn") |>
  set_mode("classification")

# fit() trains the final model on the whole training set.
knn_fit <- workflow() |>
  add_recipe(mm_recipe) |>
  add_model(knn_spec) |>
  fit(data = mm_train)

# predict() makes a finish prediction for every test team,
# and bind_cols() attaches those predictions to the test data.
mm_test_predictions <- predict(knn_fit, mm_test) |>
  bind_cols(mm_test)

# metrics() compares predictions (.pred_class) to the true finish; we keep accuracy.
mm_test_predictions |>
  metrics(truth = tourney_finish, estimate = .pred_class) |>
  filter(.metric == "accuracy")

# conf_mat() shows, for each true finish, which finishes the model predicted.
mm_test_predictions |>
  conf_mat(truth = tourney_finish, estimate = .pred_class)
```
